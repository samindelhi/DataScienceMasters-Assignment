- Q7. What do you understand about mutable and immutable data types? Give examples for both showing
this property.'''

  In Python, mutability refers to whether an object's value can be changed after it has been created viz. lists.
  
  Where the value cannot be changed after creation, that object is Immutable in nature viz. tuples.

  - Lists are mutable due its Dynamic Memory Allocation. They are stored as referece to objects due to which their element can be modified without creating new list objects.
  
  - Tuples are immmutable as they have Static or Fixed memory Allocation meaning, they store the value directly in the location instead of referencing which makes it difficult to change after creation / allocation.
